```python
# 🟦 train_test_split → Data ko Train & Test mein divide karta hai.
from sklearn.model_selection import train_test_split

# 🟩 StandardScaler → Numerical data ko same scale par laata hai.
from sklearn.preprocessing import StandardScaler

# 🟨 OneHotEncoder → Categorical data ko 0/1 format mein convert karta hai.
from sklearn.preprocessing import OneHotEncoder

# 🟧 LogisticRegression → Classification prediction karta hai.
from sklearn.linear_model import LogisticRegression

# 🟪 Pipeline → Preprocessing + Model ko ek workflow mein combine karta hai.
from sklearn.pipeline import Pipeline

# 🟥 ColumnTransformer → Different columns par different preprocessing apply karta hai.
from sklearn.compose import ColumnTransformer

# 🔵 confusion_matrix → Correct/incorrect predictions ko matrix mein show karta hai.
from sklearn.metrics import confusion_matrix

# 🟢 classification_report → Precision, Recall, F1-score deta hai.
from sklearn.metrics import classification_report


# ⭐ Overall ML Workflow:
# Preprocess → Train → Predict → Evaluate
```

```python
| Part                  | Simple Meaning                                                          |
| --------------------- | ----------------------------------------------------------------------- |
| **Pipeline**          | ML ke multiple steps ko **ek single workflow** mein combine karti hai.  |
| **Flow**              | **Raw Data → Preprocessing → Model → Prediction**                       |
| **Preprocessor**      | `StandardScaler` + `OneHotEncoder`                                      |
| **Classifier**        | `LogisticRegression`                                                    |
| **Benefit**           | Sabhi steps automatically **sequence mein execute** hote hain.          |
| **Example**           | `lr_pipeline.fit(X_train, y_train)` → Model train                       |
| **Prediction**        | `lr_pipeline.predict(X_test)` → Prediction                              |
| **Simple Definition** | **Pipeline = Preprocessing + Model ko ek workflow mein connect karna.** |
```

```python
| Code                               | Simple Meaning                                       |
| ---------------------------------- | ---------------------------------------------------- |
| `lr_pipeline.predict(X_test)`      | Test data ki **prediction** karta hai                |
| `y_pred`                           | Model ki predicted values store karta hai            |
| `confusion_matrix(y_test, y_pred)` | **Actual vs Predicted** compare karta hai            |
| `classification_report(...)`       | **Precision, Recall, F1-score** deta hai             |
| `recall_score(...)`                | Actual fraud mein se **kitne fraud detect hue**      |
| `precision_score(...)`             | Predicted fraud mein se **kitne actually fraud the** |

### 🔄 Overall Flow

**X_test → Prediction → Actual se Compare → Performance Check**

🧠 **Main purpose:** Model ne fraud ko **kitna accurately detect kiya**, ye check karna.
```

```python
| Code                                          | Simple Meaning                                                               |
| --------------------------------------------- | ---------------------------------------------------------------------------- |
| `RandomForestClassifier`                      | Random Forest classification model import karta hai                          |
| `n_estimators=100`                            | **100 decision trees** banata hai                                            |
| `class_weight="balanced"`                     | Imbalanced data mein **minority class (Fraud)** ko extra importance deta hai |
| `random_state=42`                             | Har run mein **same/reproducible result** maintain karta hai                 |
| `("preprocessor", preprocessor)`              | Pehle **Scaling + OneHotEncoding** karega                                    |
| `("classifier", RandomForestClassifier(...))` | Preprocessed data ko **Random Forest** mein bhejega                          |
| `rf_pipeline.fit(X_train,y_train)`            | Complete pipeline ko **training data par train** karta hai                   |
```
