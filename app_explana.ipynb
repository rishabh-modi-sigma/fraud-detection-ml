# 🛡️ Fraud Detection Model

```python
Model: Machine Learning model trained on transaction data.Use: Transaction Fraud / Normal detect karna.
Input Type, Amount, Sender/Receiver balance etc.
Work: Training data se fraud patterns learn karta hai.
Prediction: New transaction ko model check karke `0` ya `1` deta hai.
`0` = Normal / Non-Fraudulent ✅
`1` = Fraudulent ⚠️
Result: App prediction ko **“Legitimate”** ya **“Fraudulent”** message mein show karti hai.

### 🔄 Flow

Transaction Details → Preprocessing → ML Model → `0/1` → Result

### 🌍 Use

Banking, UPI, Cards, Online Payments & FinTech mein fraud detection ke liye.
```

# 🧠 Simple Example

```python
Type: TRANSFER
Amount: ₹5,00,000
Sender Before: ₹5,00,000
Sender After: ₹0
Receiver Before: ₹0
Receiver After: ₹0
```

Result:

⚠️ Fraudulent Transaction

Because model ne training data se similar fraud patterns learn kiye hain.

Normal example: Prediction = 0 → ✅ Legitimate / Normal

# 2️⃣ User Transaction Details Enter Karta Hai

```python
| Input                       | Meaning                                               |
| --------------------------- | ----------------------------------------------------- |
| Transaction Type        | Transaction ka type, e.g. `PAYMENT`                   |
| Transaction Amount      | Kitne ₹ ka transaction hai                            |
| Sender Balance Before   | Sender ke paas transaction se pehle kitna balance tha |
| Sender Balance After    | Transaction ke baad sender ka balance                 |
| Receiver Balance Before | Receiver ke paas pehle kitna balance tha              |
| Receiver Balance After  | Transaction ke baad receiver ka balance               |
```

```python
Example:
Transaction Type      → PAYMENT
Amount                → ₹1,000
Sender Before         → ₹10,000
Sender After          → ₹9,000
Receiver Before       → ₹5,000
Receiver After        → ₹6,000
```

Preprocessing

Model ke training time par jo preprocessing use hui thi, wahi input par apply hoti hai:

Then processed data ML model ko diya jata hai.

# 4️⃣ 🧠 Model Prediction Kaise Karta Hai?

Model ne training ke time bahut saare transaction examples se patterns learn kiye hote hain.

```python
Training Data
     ↓
Normal Transactions
Fraud Transactions
     ↓
Patterns Learn
     ↓
Trained Model
```

Ab new transaction aata hai:

```python
₹1,000
Sender Balance = ₹10,000
Receiver Balance = ₹5,000
Type = PAYMENT
        ↓
Model
        ↓
0 = Normal
1 = Fraud
```
